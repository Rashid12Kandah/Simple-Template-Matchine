```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
from skimage.metrics import mean_squared_error as mse
```

```python
img = cv2.imread('veget.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print("Image Shape: ",img.shape)
```

```python
banana = img[0:420,1500:1920]
brocoli = img[421:840,481:900]
```

```python
cv2_imshow(banana)
cv2_imshow(brocoli)
```

```python
def sliding_window(image, step, windowSize):
    for y in range(0, image.shape[0], step):
        for x in range(0, image.shape[1], step):
            yield (x, y, image[y:y + windowSize[1], x:x + windowSize[0]])
```

```python
def compare_images(imageA, imageB):
    error = mse(imageA, imageB)
    return error
```

```python
step = 50
ban_t_height, ban_t_width = banana.shape

result_banana = []
for (x, y, window) in sliding_window(img, step = step, windowSize = (ban_t_height, ban_t_width)):
    if window.shape[0] != ban_t_height or window.shape[1] != ban_t_width:
        continue


    clone = img.copy()
    cv2.rectangle(clone, (x, y), (x + ban_t_width, y + ban_t_height), (0, 255, 0), 2)

    cv2.namedWindow("Template Matching (Pear)", cv2.WINDOW_NORMAL)
    cv2.imshow("Template Matching (Pear)", clone)
    cv2.waitKey(1)

    error = compare_images(banana, window)
    result_banana.append((error, x, y))

cv2.destroyAllWindows()


result_banana.sort()
(x, y) = result_banana[0][1], result_banana[0][2]
cv2.rectangle(img, (x, y), (x + ban_t_width, y + ban_t_height), (0, 255, 0), 2)
imageText = img.copy()
cv2.putText(imageText, "Banana", (x+180,y-20), fontFace = cv2.FONT_HERSHEY_SIMPLEX, fontScale = 1, color = (0,255,0), thickness=2)
cv2.namedWindow("Detected Region", cv2.WINDOW_NORMAL)
cv2.imshow("Detected Region", imageText)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
step = 50
broc_t_height, broc_t_width = brocoli.shape

result_brocli = []
for (x, y, window) in sliding_window(img, step = step, windowSize = (broc_t_height, broc_t_width)):
    if window.shape[0] != broc_t_height or window.shape[1] != broc_t_width:
        continue


    clone = img.copy()
    cv2.rectangle(clone, (x, y), (x + broc_t_width, y + broc_t_height), (0, 255, 0), 2)

    cv2.namedWindow("Template Matching (Brocoli)", cv2.WINDOW_NORMAL)
    cv2.imshow("Template Matching (Brocoli)", clone)
    cv2.waitKey(1)

    error = compare_images(brocoli, window)
    result_brocli.append((error, x, y))

cv2.destroyAllWindows()


result_brocli.sort()
(x, y) = result_brocli[0][1], result_brocli[0][2]
cv2.rectangle(img, (x, y), (x + broc_t_width, y + broc_t_height), (0, 255, 0), 2)
imageText = img.copy()
cv2.putText(imageText, "Brocli", (x+180,y-20), fontFace = cv2.FONT_HERSHEY_SIMPLEX, fontScale = 1, color = (0,255,0), thickness=2)
cv2.namedWindow("Detected Region", cv2.WINDOW_NORMAL)
cv2.imshow("Detected Region", imageText)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
